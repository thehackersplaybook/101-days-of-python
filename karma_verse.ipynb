## Step 1: Environment Setup and Installation

This cell handles initial setup for the notebook:

- Installs dependencies from `requirements/karma_verse.requirements.txt`.

- Retries installation up to 3 times on failure.

- Loads environment variables from `.env` using `python-dotenv`.

- Ensures `OPENAI_API_KEY` is set before continuing.

After setup, it clears the output and confirms success.


```python
# Boilerplate: This block goes into every notebook.
# It sets up the environment, installs the requirements, and checks for the required environment variables.

from IPython.display import clear_output
from dotenv import load_dotenv
import os

requirements_installed = False
max_retries = 3
retries = 0
REQUIRED_ENV_VARS = ["OPENAI_API_KEY"]
PROJECT_NAME = "karma_verse"
REQUIREMENTS_FILE = f"{PROJECT_NAME}.requirements.txt"


def install_requirements():
    """Installs the requirements from requirements.txt file"""
    global requirements_installed, retries, max_retries
    if requirements_installed:
        print("Requirements already installed.")
        return

    print("Installing requirements...")
    install_status = os.system(f"pip install -r requirements/{REQUIREMENTS_FILE}")
    if install_status == 0:
        print("Requirements installed successfully.")
        requirements_installed = True
    else:
        print("Failed to install requirements.")
        if retries < max_retries:
            print("Retrying...")
            retries += 1
            return install_requirements()
        exit(1)
    return


def setup_env():
    """Sets up the environment variables"""

    def check_env(env_var):
        value = os.getenv(env_var)
        if value is None:
            print(f"Please set the {env_var} environment variable.")
            exit(1)
        else:
            print(f"{env_var} is set.")

    load_dotenv(override=True)

    variables_to_check = REQUIRED_ENV_VARS

    for var in variables_to_check:
        check_env(var)


install_requirements()
clear_output()
setup_env()
print("🚀 Setup complete. Continue to the next cell.")
```

## Step 3: Defining Data Models and Agent Configuration

This code defines Pydantic models for input and output data representing different karma analysis aspects. It sets constants for model names and verbosity.

It configures multiple specialized agents with names, instructions, model choices, and input/output types for modular karma evaluation.

- **Key Features:**

  - Defines `KarmaInput` and various output models like intention, ripple effects, karma score, emotions, journal prompts, and more.
  
  - Sets model constants (`gpt-4o`, `gpt-4o-mini`) and a global verbose flag.
  
  - Combines outputs into a single `KarmaSystemOutput` model.
  
  - Maps agents in `KARMA_AGENT_CONFIG` with specific roles, models, and data types for targeted analysis.


```python
from pydantic import BaseModel
from typing import List

LLM_MODEL_GPT_4o = "gpt-4o"
LLM_MODEL_GPT_4o_mini = "gpt-4o-mini"
GLOBAL_VERBOSE_ENABLED = True

class KarmaInput(BaseModel):
    personal_info: List[str]
    context: str

class IntentionOutput(BaseModel):
    intention_analysis: str

class RippleOutput(BaseModel):
    ripple_effects: str

class KarmaOutput(BaseModel):
    karma_score: float
    feedback: str

class PurifierOutput(BaseModel):
    purification_methods: List[str]

class EmotionOutput(BaseModel):
    emotional_states: List[str]

class JournalOutput(BaseModel):
    reflection_prompt: str

class ScriptureOutput(BaseModel):
    quote: str
    source: str

class WitnessOutput(BaseModel):
    observer_view: str

class FutureImpactOutput(BaseModel):
    karmic_forecast: str

class ReportOutput(BaseModel):
    report_markdown: str

class KarmaSystemInput(BaseModel):
    input: KarmaInput

class KarmaSystemOutput(BaseModel):
    intention: IntentionOutput
    ripple: RippleOutput
    karma: KarmaOutput
    purifier: PurifierOutput
    emotion: EmotionOutput
    journal: JournalOutput
    scripture: ScriptureOutput
    witness: WitnessOutput
    future_impact: FutureImpactOutput
    report: ReportOutput

KARMA_AGENT_CONFIG = {
    "INTENTION_AGENT": {
        "name": "Intention Agent",
        "instructions": "Analyze the motivations and thought process behind the user's action.",
        "model": LLM_MODEL_GPT_4o_mini,
        "input_type": KarmaInput,
        "output_type": IntentionOutput,
        "verbose": GLOBAL_VERBOSE_ENABLED,
    },
    "RIPPLE_AGENT": {
        "name": "Ripple Agent",
        "instructions": "Assess the emotional, social, and relational consequences of the action.",
        "model": LLM_MODEL_GPT_4o,
        "input_type": KarmaInput,
        "output_type": RippleOutput,
        "verbose": GLOBAL_VERBOSE_ENABLED,
    },
    "KARMA_AGENT": {
        "name": "Karma Judgement Agent",
        "instructions": "Evaluate the moral and spiritual implications from a karmic perspective.",
        "model": LLM_MODEL_GPT_4o,
        "input_type": KarmaInput,
        "output_type": KarmaOutput,
        "verbose": GLOBAL_VERBOSE_ENABLED,
    },
    "PURIFIER_AGENT": {
        "name": "Purifier Agent",
        "instructions": "Suggest practices or rituals to purify karmic weight.",
        "model": LLM_MODEL_GPT_4o,
        "input_type": KarmaInput,
        "output_type": PurifierOutput,
        "verbose": GLOBAL_VERBOSE_ENABLED,
    },
    "EMOTION_AGENT": {
        "name": "Emotion Agent",
        "instructions": "Detect and explain the emotional states involved.",
        "model": LLM_MODEL_GPT_4o_mini,
        "input_type": KarmaInput,
        "output_type": EmotionOutput,
        "verbose": GLOBAL_VERBOSE_ENABLED,
    },
    "JOURNAL_AGENT": {
        "name": "Journal Agent",
        "instructions": "Craft a reflection prompt for introspection.",
        "model": LLM_MODEL_GPT_4o_mini,
        "input_type": KarmaInput,
        "output_type": JournalOutput,
        "verbose": GLOBAL_VERBOSE_ENABLED,
    },
    "SCRIPTURE_AGENT": {
        "name": "Scripture Agent",
        "instructions": "Provide a spiritual quote relevant to the context.",
        "model": LLM_MODEL_GPT_4o,
        "input_type": KarmaInput,
        "output_type": ScriptureOutput,
        "verbose": GLOBAL_VERBOSE_ENABLED,
    },
    "WITNESS_AGENT": {
        "name": "Witness Agent",
        "instructions": "Describe how a neutral spiritual observer might interpret the action.",
        "model": LLM_MODEL_GPT_4o,
        "input_type": KarmaInput,
        "output_type": WitnessOutput,
        "verbose": GLOBAL_VERBOSE_ENABLED,
    },
    "FUTURE_IMPACT_AGENT": {
        "name": "Future Impact Agent",
        "instructions": "Predict potential long-term karmic and emotional effects.",
        "model": LLM_MODEL_GPT_4o,
        "input_type": KarmaInput,
        "output_type": FutureImpactOutput,
        "verbose": GLOBAL_VERBOSE_ENABLED,
    },
    "REPORT_AGENT": {
        "name": "Report Agent",
        "instructions": "Generate a structured and well formatted markdown report using all the karma agent insights. Give a detailed analysis of the user's action, including the intention, ripple effects, karmic implications, purification methods, emotional states, reflection prompts, spiritual insights, neutral observations, and future impacts.",
        "model": LLM_MODEL_GPT_4o,
        "input_type": KarmaInput,
        "output_type": ReportOutput,
        "verbose": GLOBAL_VERBOSE_ENABLED,
    },
}
```

## Step 4: Running Karma Agents with OpenAI API

This code initializes an OpenAI client and defines a function `run_agent` to run any karma agent from the configuration.

- It takes an `agent_key`, `personal_info`, and `context` as inputs.

- Validates the agent key against `KARMA_AGENT_CONFIG`.

- Builds a detailed prompt using the agent’s name, instructions, user info, and context.

- Sends the prompt to the OpenAI chat completion API with the specified model and low temperature for focused output.

- Returns the agent’s text response or an error message if the call fails.


```python
from openai import OpenAI

client = OpenAI(api_key=os.getenv("OPENAI_API_KEY"))

def run_agent(agent_key: str, personal_info: List[str], context: str) -> str:
    """
    Runs a specified karma agent with the provided personal information and context.

    Args:
        agent_key (str): The key of the agent to run.
        personal_info (List[str]): Personal information about the user.
        context (str): Context of the action or situation.

    Returns:
        str: The response from the agent.
    """
    if agent_key not in KARMA_AGENT_CONFIG:
        raise ValueError(f"Agent key '{agent_key}' not found in configuration.")

    config = KARMA_AGENT_CONFIG[agent_key]
    prompt = f"""
You are {config['name']}, an expert agent.

Task Instructions:
{config['instructions']}

User Personal Information:
{personal_info}

Context of the Action or Situation:
{context}

Please provide a detailed and thoughtful response based on the above information.
""".strip()
    
    try:
        response = client.chat.completions.create(
            model=config["model"],
            messages=[{"role": "user", "content": prompt}],
            temperature=0.2,
        )
        text = response.choices[0].message.content.strip()
        return text
    except Exception as e:
        print(f"❌ Error running agent {agent_key}: {e}")
        return f"[Error running {config['name']}]"
```

## Step 5: Aggregating Karma Agent Outputs into a Comprehensive Report

This function `karma_mirror` runs all karma agents sequentially with the given personal info and context.


- Calls each specialized agent to get insights on intention, ripple effects, karma, purification, emotions, reflection prompts, scriptures, witness view, and future impact.

- Combines these individual outputs into a detailed markdown report template.

- Finally, passes the combined report context to the `REPORT_AGENT` for a polished, structured summary.

- Returns the final karma mirror report as a markdown string.



```python
from typing import List

def karma_mirror(personal_info: List[str], context:str) -> str:
    """
    Run the Karma Mirror Analysis on the provided personal information and context.

    Args:
        personal_info (List[str]): List of personal information about the user.
        context (str): Context of the action or situation to analyze.

    Returns:
        str: A structured markdown report summarizing the analysis.
    """
    print("🧘 Running Karma Mirror Analysis...\n")

    intention = run_agent("INTENTION_AGENT", personal_info, context)
    ripple = run_agent("RIPPLE_AGENT", personal_info, context)
    karma = run_agent("KARMA_AGENT", personal_info, context)
    purifier = run_agent("PURIFIER_AGENT", personal_info, context)
    emotion = run_agent("EMOTION_AGENT", personal_info, context)
    journal = run_agent("JOURNAL_AGENT", personal_info, context)
    scripture = run_agent("SCRIPTURE_AGENT", personal_info, context)
    witness = run_agent("WITNESS_AGENT", personal_info, context)
    future_impact = run_agent("FUTURE_IMPACT_AGENT", personal_info, context)

    combined_context = f"""
### 🪞 Karma Mirror Report

**👤 Personal Info:** `{', '.join(personal_info)}`
**📜 Context:** _{context}_

---

#### 🎯 Intention Analysis
> {intention}

#### 🌊 Ripple Effects
> {ripple}

#### ⚖️ Karma Judgement
> {karma}

#### 🔥 Purification Suggestions
> {purifier}

#### 💓 Emotional States
> {emotion}

#### ✍️ Reflection Prompt
> {journal}

#### 📖 Spiritual Quote
> {scripture}

#### 👁️ Observer View
> {witness}

#### 🧭 Future Impact
> {future_impact}
"""

    report = run_agent("REPORT_AGENT", personal_info, combined_context)
    return report
```

## Step 6: Displaying the Karma Mirror Report in Markdown

This code snippet calls the `karma_mirror` function with empty inputs and displays the resulting report as formatted Markdown using IPython's `display` and `Markdown`.

- Initializes empty `personal_info` and `context` variables.

- Runs the karma analysis pipeline to generate the final report.

- Uses `display(Markdown(final_response))` to render the report cleanly in Jupyter or similar environments.


```python
from IPython.display import Markdown, display

personal_info = []
context = ""

final_response = karma_mirror(personal_info, context)

display(Markdown(final_response))
```
